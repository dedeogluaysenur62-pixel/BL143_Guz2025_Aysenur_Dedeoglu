```python
def gun_planlayici():
    while True:
        try:
            # 1, 2 ve 3. Maddeler: Girdi alma
            uyku = float(input("Günlük uyku süresini giriniz (saat): "))
            ders = float(input("Günlük ders süresini giriniz (saat): "))
            serbest = float(input("Günlük serbest zaman süresini giriniz (saat): "))

            # 5. Madde: Negatif değer kontrolü
            if uyku < 0 or ders < 0 or serbest < 0:
                print("Hata: Süreler negatif olamaz! Lütfen tekrar deneyin.\n")
                continue

            # 6. Madde: 24 saat kontrolü
            toplam = uyku + ders + serbest
            if toplam > 24:
                print(f"Hata: Toplam süre ({toplam} saat) 24 saati geçemez! Tekrar giriniz.\n")
                continue

            # Geçerli girişler yapıldıysa döngüden çık
            break

        except ValueError:
            # 4. Madde: Sayı olup olmadığını kontrol et
            print("Hata: Lütfen sadece sayısal değerler giriniz!\n")

    # 7. Madde: Uyku durumu belirleme
    if uyku < 6:
        uyku_durumu = "Yetersiz uyku"
    elif 6 <= uyku <= 8:
        uyku_durumu = "Normal uyku"
    else:
        uyku_durumu = "Fazla uyku"

    # 8. Madde: Çalışma yoğunluğu belirleme
    if ders >= 6:
        calisma_durumu = "Yoğun çalışma günü"
    else:
        calisma_durumu = "Normal çalışma günü"

    # 9. Madde: Genel değerlendirme
    if uyku < 6 and ders >= 6:
        genel_sonuc = "Verimsiz gün"
    else:
        genel_sonuc = "Dengeli gün"

    # Çıktı Formatı
    print("\n--- SONUÇLAR ---")
    print(f"Uyku durumu: {uyku_durumu}")
    print(f"Çalışma durumu: {calisma_durumu}")
    print(f"Genel değerlendirme: {genel_sonuc}")

# Fonksiyonu çalıştır
gun_planlayici()
```

```python
Değişken: Bilgisayar hafızasında verileri saklamak için kullanılan isimlendirilmiş alanlardır. Örnek: uyku = 7.
Koşul: Programın belirli şartlara göre farklı yollar izlemesini sağlayan yapılardır. if, elif ve else anahtar kelimeleriyle kurulur.
Karşılaştırma Operatörü: İki değeri birbiriyle kıyaslamak için kullanılır. Örn: > (büyüktür), < (küçüktür), == (eşittir), >= (büyük eşittir).
Girdi / Çıktı: Kullanıcıdan veri alma işlemine Girdi (input()), işlenen veriyi ekrana yazdırma işlemine Çıktı (print()) denir.
Hata Kontrolü: Programın çalışma sırasında beklenmedik durumlarla (yanlış veri girişi gibi) karşılaştığında çökmesini engellemek için kullanılan yapılardır. Örnek: try-except bloğu.
```
